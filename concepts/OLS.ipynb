# What is OLS
## Do you know the basic math behind your algorithms, well you should

## OLS Theory

Ordinary Least Squares is a method for choosing our parameters in a regression model. To understand this algorithm we first need to understand the parameters that we are choosing. Let's take the following example:

We are trying to predict a person's height given their weight. Given the height and weight of a person we can say:

y = mx + b
where y is a person's height and x is a person's weight: 
diabetes_progression = m * weight + b

OLS is an algorith to find the the slope an intercept that will minimize the square residuals. We do so by following these steps:

1. Fit a line to the data
2. Calculate the sum of squared residuals
3. Adjust and repeat!

## So what exactly is R2?
R2 is the thing we are trying to minimize in OLS, so what is it? R2 is formally defined as:

### $R^{2} = 1 - \frac{SS_{res}}{SS_{total}} $

### $SS_{res} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

### $SS_{total} = \sum_{i=1}^{n} (y_i - \bar{y})^2$

Let's break the equation down. The total sum of squares (SS_total) is simply the variance around the mean


1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}$

```python
# create dummy regression problem
data, target = make_regression(5, 2, bias=0.5, noise=10)
X = pd.DataFrame(data, columns = ["weight", "shoe_size"])
y = pd.DataFrame(target, columns=["height"])
df = pd.concat([X, y], axis=1)
```

```python
# Fit an OLS regression model to only weight
ols = LinearRegression()
ols.fit(X[["weight"]], y)
df["y_hat"] = ols.predict(X[["weight"]])
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(name="y_true", mode='markers', x=df["weight"], y=df["height"]))
fig.add_trace(go.Scatter(mode='lines', name="Mean", x=df["weight"], y=mean_y, line=dict(color='green')))
for i, (weight, y, y_hat) in enumerate(var_mean):
    if (i == 0):
        fig.add_trace(go.Scatter(name="residuals", x=[weight, weight], y=[y, y_hat], mode='lines', line=dict(dash='dash', color="red"), showlegend=True))
    else:
        fig.add_trace(go.Scatter(name="residuals", x=[weight, weight], y=[y, y_hat], mode='lines', line=dict(dash='dash', color="red"), showlegend=False))

fig.update_layout(
    title="Total Sum of Square Visualization",
    width=700,
    height=500
)
fig.show()
```

## Sum of Square Residuals Calculation

### $SS_{res} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$



```python
# Sum of Square Residuals

SS_res = ((df["height"] - df["y_hat"]) ** 2).sum()
SS_res
```

## Sum of Square Total Calculation

### $SS_{total} = \sum_{i=1}^{n} (y_i - \bar{y})^2$


```python
SS_total = ((df["height"] - (df["height"].mean())) ** 2).sum()
SS_total
```

```python
1 - SS_res / SS_total
```

```python
# with scikit learn
r2_score(df["height"], df["y_hat"])
```
